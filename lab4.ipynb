```python


import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder

# Step 1: Create a sample dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [20, np.nan, 22, 21, np.nan],
    'Marks': [85, 90, np.nan, 75, 80],
    'City': ['Delhi', 'Mumbai', np.nan, 'Delhi', 'Mumbai']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Mean Imputation
mean_imputer = SimpleImputer(strategy='mean')
df[['Age']] = mean_imputer.fit_transform(df[['Age']])

print("\nAfter Mean Imputation (Age):")
print(df)

# Step 3: Median Imputation
median_imputer = SimpleImputer(strategy='median')
df[['Marks']] = median_imputer.fit_transform(df[['Marks']])

print("\nAfter Median Imputation (Marks):")
print(df)

# Step 4: Forward Fill
df['City'] = df['City'].ffill()

print("\nAfter Forward Fill (City):")
print(df)

# Step 5: Encoding Categorical Variables
label_encoder = LabelEncoder()
df['City_Encoded'] = label_encoder.fit_transform(df['City'])

print("\nAfter Encoding Categorical Variables:")
print(df)

# Step 6: Display Final Dataset
print("\nFinal Preprocessed Dataset:")
print(df)
```
