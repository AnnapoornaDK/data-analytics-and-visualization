```python

# Feature Engineering, Normalization, Standardization and PCA

# Step 1: Import Libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Step 2: Load Your CSV Dataset
df = pd.read_csv("students.csv")

print("Original Dataset:")
print(df.head())

# Step 3: Feature Engineering
# Select numerical columns
features = df.select_dtypes(include=[np.number]).copy()

# Handle missing values
features = features.fillna(features.mean())

# Create a new feature using the first two numerical columns
if features.shape[1] >= 2:
    features['New_Feature'] = (
        features.iloc[:, 0] * features.iloc[:, 1]
    )

print("\nAfter Feature Engineering:")
print(features.head())

# Step 4: Min-Max Normalization (0 to 1)
minmax = MinMaxScaler()

normalized = minmax.fit_transform(features)

normalized_df = pd.DataFrame(
    normalized,
    columns=features.columns
)

print("\nMin-Max Normalization:")
print(normalized_df.head())

# Step 5: Z-Score Standardization (Mean=0, Std=1)
scaler = StandardScaler()

standardized = scaler.fit_transform(features)

standardized_df = pd.DataFrame(
    standardized,
    columns=features.columns
)

print("\nZ-Score Standardization:")
print(standardized_df.head())

# Step 6: PCA - Dimensionality Reduction
if standardized.shape[1] >= 2:
    pca = PCA(n_components=2)

    principal_components = pca.fit_transform(standardized)

    pca_df = pd.DataFrame(
        principal_components,
        columns=['PC1', 'PC2']
    )

    print("\nPCA - Principal Components:")
    print(pca_df.head())

    # Step 7: Explained Variance
    print("\nExplained Variance Ratio:")
    print(pca.explained_variance_ratio_)

    print("\nTotal Explained Variance:")
    print(sum(pca.explained_variance_ratio_))

else:
    print("\nPCA requires at least 2 numerical features.")

# Step 8: Final Output
print("\nProgram Executed Successfully!")
```

```python

import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# STEP 1: Enter Student Dataset Directly

data = {
    'StudentID': ['S1000','S1001','S1002','S1003','S1004',
                  'S1005','S1006','S1007','S1008','S1009',
                  'S1010','S1011','S1012','S1013','S1014'],

    'StudyHours': [5.08,6.16,5.42,2.93,8.01,7.15,7.24,3.5,
                   8.59,2.42,6.79,10.32,3.32,4.25,5.72],

    'Attendance': [82.32,68.27,98.16,79.5,93.33,78.88,100,
                   100,74.76,90.63,86.39,95.79,65.46,86.92,91.76],

    'PrevMarks': [65.44,82.55,67.71,68.25,60.55,65.15,70.62,
                  40.35,45.78,77.15,68.56,63.24,66.28,71.21,57.9],

    'Marks': [82.46,79.14,76.89,62.16,86.49,83.57,91.56,
              63.71,81.96,78.22,91.43,100,62.04,73.45,80.52],

    'Branch': ['ISE','CSD','CSE','ISE','ISE','CSD','CSE',
               'CSE','ISE','AIML','ISE','ISE','ISE','ISE','CSD'],

    'Result': ['Pass','Pass','Pass','Fail','Pass','Pass','Pass',
               'Fail','Pass','Pass','Pass','Pass','Fail','Pass','Pass']
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET")
print(df)

# STEP 2: CHECK MISSING VALUES

print("\nMISSING VALUES")
print(df.isnull().sum())

# STEP 3: REMOVE DUPLICATES

df = df.drop_duplicates()

print("\nDUPLICATES REMOVED")
print(df.shape)

# STEP 4: OUTLIER DETECTION USING IQR

numeric_columns = ['StudyHours', 'Attendance', 'PrevMarks', 'Marks']

print("\nOUTLIER DETECTION")

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("\nColumn:", col)
    print("Outliers:", len(outliers))
    print(outliers[col].tolist())

# STEP 5: Z-SCORE CALCULATION

z_scores = (
    df[numeric_columns] - df[numeric_columns].mean()
) / df[numeric_columns].std(ddof=0)

print("\nZ-SCORE VALUES")
print(z_scores)

print("\nZ-SCORE OUTLIERS")
print((z_scores.abs() > 3).sum())

# STEP 6: HANDLE MISSING VALUES

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)

# STEP 7: FEATURE ENGINEERING

df['Study_Attendance_Score'] = (
    df['StudyHours'] * df['Attendance'] / 100
)

df['Performance_Score'] = (
    df['PrevMarks'] + df['Marks']
) / 2

print("\nFEATURE ENGINEERING")
print(df.head())

# STEP 8: MIN-MAX NORMALIZATION

features = df[
    ['StudyHours', 'Attendance', 'PrevMarks', 'Marks',
     'Study_Attendance_Score', 'Performance_Score']
]

minmax = MinMaxScaler()

normalized = minmax.fit_transform(features)

normalized_df = pd.DataFrame(
    normalized,
    columns=features.columns
)

print("\nMIN-MAX NORMALIZATION (0 TO 1)")
print(normalized_df)

# STEP 9: Z-SCORE STANDARDIZATION

scaler = StandardScaler()

standardized = scaler.fit_transform(features)

standardized_df = pd.DataFrame(
    standardized,
    columns=features.columns
)

print("\nSTANDARDIZATION (MEAN 0, STD 1)")
print(standardized_df)

# STEP 10: PCA DIMENSIONALITY REDUCTION

pca = PCA(n_components=2)

principal_components = pca.fit_transform(standardized)

pca_df = pd.DataFrame(
    principal_components,
    columns=['PC1', 'PC2']
)

print("\nPCA PRINCIPAL COMPONENTS")
print(pca_df)

# STEP 11: EXPLAINED VARIANCE

print("\nEXPLAINED VARIANCE RATIO")
print(pca.explained_variance_ratio_)

print("\nTOTAL EXPLAINED VARIANCE")
print(sum(pca.explained_variance_ratio_))

# STEP 12: SAVE OUTPUT FILES

df.to_csv("cleaned_student_data.csv", index=False)
normalized_df.to_csv("normalized_student_data.csv", index=False)
standardized_df.to_csv("standardized_student_data.csv", index=False)
pca_df.to_csv("pca_student_data.csv", index=False)

print("\nPROGRAM EXECUTED SUCCESSFULLY!")
```

```python

# ============================================
# WORLD ECONOMIC INDICATORS DATA ANALYSIS
# Data Cleaning, Missing Values, Outliers,
# Z-Score, Normalization, Standardization and PCA
# ============================================

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA


# STEP 2: LOAD DATASET

df = pd.read_csv('World Economic Indicators - Sheet1.csv')

print("ORIGINAL DATASET")
print(df.head())

print("\nDataset Shape:")
print(df.shape)


# STEP 3: SELECT REQUIRED FEATURES

features = [
    'Infant mortality rate per 1000 live births',
    'Life expectancy at birth',
    'Population density',
    'Unemployment'
]

data = df[features].copy()


# STEP 4: CHECK MISSING VALUES

print("\nMISSING VALUES:")
print(data.isnull().sum())

# Replace infinite values with NaN
data = data.replace([np.inf, -np.inf], np.nan)

# Fill missing values with column mean
data = data.fillna(data.mean())

print("\nAFTER HANDLING MISSING VALUES:")
print(data.isnull().sum())


# STEP 5: REMOVE DUPLICATES

print("\nDUPLICATE RECORDS:")
print(df.duplicated().sum())

df = df.drop_duplicates()

data = df[features].copy()

data = data.replace([np.inf, -np.inf], np.nan)
data = data.fillna(data.mean())

print("\nAFTER DUPLICATE REMOVAL:")
print(data.shape)


# STEP 6: OUTLIER DETECTION USING IQR

print("\nOUTLIER DETECTION:")

for col in features:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[col] < lower) |
        (data[col] > upper)
    ]

    print("\nColumn:", col)
    print("Number of Outliers:", len(outliers))
    print(outliers[col].tolist())


# STEP 7: Z-SCORE CALCULATION

print("\nZ-SCORE CALCULATION:")

z_scores = (
    data - data.mean()
) / data.std(ddof=0)

print(z_scores.head())

print("\nZ-SCORE OUTLIERS:")
print((z_scores.abs() > 3).sum())


# STEP 8: FEATURE ENGINEERING

data['Health_Index'] = (
    data['Life expectancy at birth'] /
    data['Infant mortality rate per 1000 live births']
)

print("\nAFTER FEATURE ENGINEERING:")
print(data.head())


# STEP 9: MIN-MAX NORMALIZATION (0 TO 1)

minmax = MinMaxScaler()

normalized = minmax.fit_transform(data)

normalized_df = pd.DataFrame(
    normalized,
    columns=data.columns
)

print("\nMIN-MAX NORMALIZATION:")
print(normalized_df.head())

print("\nNormalized Minimum:", normalized.min())
print("Normalized Maximum:", normalized.max())


# STEP 10: Z-SCORE STANDARDIZATION

scaler = StandardScaler()

standardized = scaler.fit_transform(data)

standardized_df = pd.DataFrame(
    standardized,
    columns=data.columns
)

print("\nSTANDARDIZED DATA:")
print(standardized_df.head())

print("\nStandardized Mean:")
print(standardized_df.mean().round(4))

print("\nStandardized Standard Deviation:")
print(standardized_df.std(ddof=0).round(4))


# STEP 11: PCA DIMENSIONALITY REDUCTION

pca = PCA(n_components=2)

principal_components = pca.fit_transform(standardized)

pca_df = pd.DataFrame(
    principal_components,
    columns=['PC1', 'PC2']
)

print("\nPCA PRINCIPAL COMPONENTS:")
print(pca_df.head())


# STEP 12: EXPLAINED VARIANCE

print("\nEXPLAINED VARIANCE RATIO:")
print(pca.explained_variance_ratio_)

print("\nTOTAL VARIANCE CAPTURED:")
print(
    sum(pca.explained_variance_ratio_) * 100,
    "%"
)


# STEP 13: PCA VISUALIZATION

plt.figure(figsize=(8, 5))

plt.scatter(
    principal_components[:, 0],
    principal_components[:, 1],
    color='blue',
    alpha=0.6
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

plt.title('PCA: 4 Features Reduced to 2 Components')

plt.grid(True)
plt.show()


# STEP 14: SAVE OUTPUT FILES

data.to_csv('cleaned_world_data.csv', index=False)

normalized_df.to_csv(
    'normalized_world_data.csv', index=False
)

standardized_df.to_csv(
    'standardized_world_data.csv', index=False
)

pca_df.to_csv(
    'pca_world_data.csv', index=False
)


# STEP 15: FINAL OUTPUT

print("\nALL OPERATIONS COMPLETED SUCCESSFULLY!")
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

df = pd.read_csv('World Economic Indicators - Sheet1.csv')
features = df[['Infant mortality rate per 1000 live births','Life expectancy at birth','Population density','Unemployment']]

# --- Min-Max Normalization (scales to 0-1) ---
minmax = MinMaxScaler()
normalized = minmax.fit_transform(features)
print('Normalized range:', normalized.min(), 'to', normalized.max())

# --- Z-score Standardization (mean=0, std=1) ---
scaler = StandardScaler()
standardized = scaler.fit_transform(features)
print('Standardized mean:', standardized.mean().round(4), 'std:', standardized.std().round(4))

# --- PCA for dimensionality reduction ---
pca = PCA(n_components=2)
principal_components = pca.fit_transform(standardized)
print('\nExplained variance ratio:', pca.explained_variance_ratio_)
print('Total variance captured by 2 components:', sum(pca.explained_variance_ratio_)*100, '%')

plt.scatter(principal_components[:,0], principal_components[:,1], c=df['Result'].map({'Pass':'green','Fail':'red'}))
plt.xlabel('PC1'); plt.ylabel('PC2'); plt.title('PCA: 4 features reduced to 2'); plt.show()
```

```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

df = pd.read_csv('World Economic Indicators - Sheet1.csv')

features = df[[
    'Infant mortality rate per 1000 live births',
    'Life expectancy at birth',
    'Population density',
    'Unemployment'
]]

# Handle missing values
features = features.apply(pd.to_numeric, errors='coerce')
features = features.fillna(features.mean())

# --- Min-Max Normalization (scales to 0-1) ---
minmax = MinMaxScaler()
normalized = minmax.fit_transform(features)

print('Normalized range:', normalized.min(), 'to', normalized.max())

# --- Z-score Standardization (mean=0, std=1) ---
scaler = StandardScaler()
standardized = scaler.fit_transform(features)

print('Standardized mean:', standardized.mean().round(4),
      'std:', standardized.std().round(4))

# --- PCA for dimensionality reduction ---
pca = PCA(n_components=2)
principal_components = pca.fit_transform(standardized)

print('\nExplained variance ratio:', pca.explained_variance_ratio_)

print('Total variance captured by 2 components:',
      sum(pca.explained_variance_ratio_) * 100, '%')

# --- PCA Visualization ---
plt.figure(figsize=(8, 5))

plt.scatter(
    principal_components[:, 0],
    principal_components[:, 1],
    color='blue'
)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA: 4 Features Reduced to 2')
plt.show()
```
