```python
import pandas as pd

# Load the dataset
df = pd.read_csv('students.csv')

# Display summary statistics
print(df.describe())

# Check for missing values and duplicate rows
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())

# Create a DataFrame for Fees and merge it with the student data
Fee = pd.DataFrame({'Branch': ['CSE', 'ECE', 'ISE'], 'Annual Fee': [120000, 110000, 115000]})
merged = pd.merge(df, Fee, on='Branch', how='left')
print(merged)

# Calculate average marks grouped by Branch
print('\nAverage Marks by Branch:\n', df.groupby('Branch')['Marks'].mean())

# Create and display a pivot table counting students by Branch and Result
pivot = pd.pivot_table(df, values='Marks', index='Branch', columns='Result', aggfunc='count')
print(pivot)
```
